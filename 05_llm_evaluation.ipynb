```python
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

df = pd.read_csv("df_bert_nv_annotated_llm.csv", sep=";")

# binarize the labels
def convert_label(x):
    if x == "original_harry_potter":
        return 1
    elif x == "fanfiction":
        return 0
    else:
        return None

# use the good column name for the labels
df["pred"] = df["llm_label"].apply(convert_label)

# supress errors
df = df.dropna(subset=["pred"])

y_true = df["categorie"]
y_pred = df["pred"]

# metrics
accuracy = accuracy_score(y_true, y_pred)
report = classification_report(y_true, y_pred)
cm = confusion_matrix(y_true, y_pred)

print("Accuracy:", accuracy)
print("\nClassification report:\n", report)
print("\nConfusion matrix:\n", cm)
```
